```python
# rera;flattype;status

# not to encode them together # pd.get_dummies(data["RERA Approval"],drop_first=True)

# Suppose you keep all three dummy columns.
# Ready + Resale + Under Construction = 1
# This creates perfect multicollinearity (the dummy variable trap). 
# The model can't uniquely estimate the coefficients because one column is an exact linear combination of the others.
# Dropping one dummy solves this.


# For Tree-Based Models (Random Forest, XGBoost)
# Tree models do not suffer from this problem.
# So:
# drop_first=True is not required for Random Forest, XGBoost, Decision Trees.
# It is still okay to use, but many practitioners simply keep all dummy columns with tree models.

# so short summary the coefficent exchange might lead to same answer
#  if all one hot features are there but with a dropped featture model understands the parameters better
# This is mainly important for Linear Regression-type models.
```

```python
# Categorical Feature Handling Summary

# 1. Low Cardinality Features
# --------------------------------
# Meaning:
# - Features with a small number of unique categories.
# - Safe to apply One-Hot Encoding directly.

# Examples in our dataset:

# RERA Approval (2 values)
# - Approved by RERA
# - Not approved by RERA

# Action:
# ✅ Keep
# ✅ One-hot encode


# Status (4 values)
# - Ready to move
# - Under Construction
# - Resale
# - ...

# Action:
# ✅ Keep
# ✅ One-hot encode


# Flat Type (6 values)
# - Apartment
# - House
# - Plot
# - Penthouse
# - ...

# Action:
# ✅ Keep
# ✅ One-hot encode


# Is_Plot (2 values)
# - Already converted into 0/1

# Action:
# ✅ Keep
# ❌ No encoding required


# ------------------------------------------------

# 2. Medium Cardinality Features
# --------------------------------
# Meaning:
# - Features with a moderate number of unique categories.
# - Need to check whether they carry useful information.
# - One-hot encoding is possible if the dataset is large enough.

# Locality (245 values)

# Reason to keep:
# - Location is one of the strongest factors affecting property price.
# - Different sectors have different price ranges.

# Action:
# ✅ Keep
# ✅ One-hot encode


# Company Name (269 values)

# Reason to keep:
# - Builder/brand reputation affects property value.
# - Examples:
#   M3M, DLF, Godrej, Sobha

# Action:
# ✅ Keep
# ✅ One-hot encode


# ------------------------------------------------

# 3. High Cardinality Features
# --------------------------------
# Meaning:
# - Features with many unique categories.
# - Direct one-hot encoding can create too many columns.
# - May increase sparsity and overfitting.

# Builder Name (725 values)

# Observation:
# - Many values are not actual builders.
# - Examples:
#   home
#   Property In Gurgaon
#   properties for sale in Gurgaon

# Problem:
# - Contains broker/agent names and noisy information.

# Action:
# ❌ Drop


# Socity (967 values)

# Observation:
# - Unlike Builder Name, Society contains useful information.
# - Examples:
#   DLF Camellias
#   M3M Capital
#   Godrej Air

# Problem:
# - Too many categories.

# Possible solutions:

# Option 1:
# One-hot encode all categories.

# Option 2 (better):
# Keep only top N societies and group the rest as "Others".

# Action:
# ⚠️ Keep
# ⚠️ Reduce categories before encoding


# ------------------------------------------------

# Final Feature Decisions

# Keep:
# ✅ RERA Approval
# ✅ Status
# ✅ Flat Type
# ✅ Is_Plot
# ✅ Locality
# ✅ Company Name
# ✅ Socity (after reducing categories)

# Drop:
# ❌ Builder Name


# Encoding Plan:

# One-hot encode:
# - RERA Approval
# - Status
# - Flat Type
# - Locality
# - Company Name
# - Socity

# Keep as numerical:
# - Area
# - Rate per sqft
# - BHK_Count
# - Is_Plot
# - log_price (target)
```
